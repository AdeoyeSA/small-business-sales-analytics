## 1. Creating a Realistic Raw Sales Dataset

For this portfolio project, I am creating a synthetic transactional dataset representing sales from a fictional Canadian small business.

Creating the dataset allows me to demonstrate the complete analytics workflow, including data-quality assessment and cleaning, without exposing any real customer information.

The dataset will contain transaction-level information such as:

### Why create a "messy" dataset?

Real-world business data is rarely perfectly structured. Before analysis, an analyst must identify and resolve data-quality issues.

Therefore, this raw dataset intentionally contains examples of:

* Duplicate transactions
* Missing values
* Inconsistent text formatting
* Extra whitespace
* Invalid dates
* Invalid numerical values

The purpose is to demonstrate the process of **profiling, cleaning and validating data before using it for business analysis.**


```python
# ============================================
# STEP 2: CREATE RAW SALES DATASET
# ============================================

import pandas as pd
import numpy as np

# Make the results reproducible
np.random.seed(42)

# Number of original transactions
n = 10000

# --------------------------------------------
# Business locations
# --------------------------------------------

locations = {
    "Alberta": ["Edmonton", "Calgary", "Red Deer", "Lethbridge"],
    "British Columbia": ["Vancouver", "Surrey", "Victoria", "Kelowna"],
    "Ontario": ["Toronto", "Ottawa", "Mississauga", "Hamilton"],
    "Manitoba": ["Winnipeg", "Brandon"],
    "Saskatchewan": ["Saskatoon", "Regina"]
}

# --------------------------------------------
# Products
# --------------------------------------------

products = [
    ("Laptop Pro 14", "Electronics", 1299),
    ("Wireless Headphones", "Electronics", 179),
    ("Office Monitor 27", "Electronics", 349),
    ("Mechanical Keyboard", "Accessories", 129),
    ("Wireless Mouse", "Accessories", 49),
    ("USB-C Dock", "Accessories", 159),
    ("Office Chair", "Furniture", 429),
    ("Standing Desk", "Furniture", 699),
    ("Desk Lamp", "Furniture", 89),
    ("Notebook Pack", "Office Supplies", 24),
    ("Printer Paper", "Office Supplies", 39),
    ("Ink Cartridge", "Office Supplies", 79),
    ("Webcam HD", "Electronics", 99),
    ("Tablet 11", "Electronics", 499),
    ("Backpack", "Accessories", 79)
]

# --------------------------------------------
# Dates
# --------------------------------------------

dates = pd.date_range(
    start="2025-01-01",
    end="2026-06-30",
    freq="D"
)

date_choices = np.random.choice(dates, n)

# --------------------------------------------
# Province and city
# --------------------------------------------

province_choices = np.random.choice(
    list(locations.keys()),
    n,
    p=[0.30, 0.18, 0.32, 0.10, 0.10]
)

city_choices = [
    np.random.choice(locations[province])
    for province in province_choices
]

# --------------------------------------------
# Product information
# --------------------------------------------

product_indices = np.random.randint(
    0,
    len(products),
    n
)

product_names = [
    products[i][0]
    for i in product_indices
]

categories = [
    products[i][1]
    for i in product_indices
]

unit_prices = np.array([
    products[i][2]
    for i in product_indices
], dtype=float)

# --------------------------------------------
# Quantity and discount
# --------------------------------------------

quantity = np.random.choice(
    [1, 2, 3, 4, 5, 6],
    n,
    p=[0.30, 0.27, 0.18, 0.12, 0.08, 0.05]
)

discount = np.random.choice(
    [0, 0.05, 0.10, 0.15, 0.20],
    n,
    p=[0.45, 0.20, 0.20, 0.10, 0.05]
)

# --------------------------------------------
# Financial calculations
# --------------------------------------------

sales = (
    quantity
    * unit_prices
    * (1 - discount)
)

cost_ratio = np.random.uniform(
    0.55,
    0.82,
    n
)

cost = sales * cost_ratio

profit = sales - cost

# --------------------------------------------
# Customers
# --------------------------------------------

customer_numbers = np.random.randint(
    1001,
    4200,
    n
)

customer_ids = [
    f"CUST-{x}"
    for x in customer_numbers
]

customer_names = [
    f"Customer {x}"
    for x in customer_numbers
]

# --------------------------------------------
# Payment and sales channel
# --------------------------------------------

payment_methods = [
    "Credit Card",
    "Debit",
    "E-Transfer",
    "PayPal"
]

sales_channels = [
    "Online",
    "Retail Store",
    "Wholesale"
]

# --------------------------------------------
# Build DataFrame
# --------------------------------------------

raw_sales = pd.DataFrame({

    "Order_ID": [
        f"ORD-{100000 + i}"
        for i in range(n)
    ],

    "Order_Date": pd.to_datetime(
        date_choices
    ).strftime("%Y-%m-%d"),

    "Customer_ID": customer_ids,

    "Customer_Name": customer_names,

    "Province": province_choices,

    "City": city_choices,

    "Product": product_names,

    "Category": categories,

    "Quantity": quantity,

    "Unit_Price": np.round(
        unit_prices,
        2
    ),

    "Discount": discount,

    "Sales": np.round(
        sales,
        2
    ),

    "Cost": np.round(
        cost,
        2
    ),

    "Profit": np.round(
        profit,
        2
    ),

    "Payment_Method": np.random.choice(
        payment_methods,
        n
    ),

    "Sales_Channel": np.random.choice(
        sales_channels,
        n,
        p=[0.55, 0.30, 0.15]
    )
})

# ============================================
# INTRODUCE REALISTIC DATA-QUALITY PROBLEMS
# ============================================

# 1. Add duplicate transactions
duplicates = raw_sales.sample(
    75,
    random_state=7
)

raw_sales = pd.concat(
    [raw_sales, duplicates],
    ignore_index=True
)

# 2. Missing values
for column, count in [
    ("City", 55),
    ("Customer_Name", 40),
    ("Discount", 30),
    ("Payment_Method", 25)
]:

    indexes = np.random.choice(
        raw_sales.index,
        count,
        replace=False
    )

    raw_sales.loc[
        indexes,
        column
    ] = np.nan

# 3. Inconsistent capitalization
indexes = np.random.choice(
    raw_sales.index,
    120,
    replace=False
)

raw_sales.loc[
    indexes[:40],
    "Province"
] = raw_sales.loc[
    indexes[:40],
    "Province"
].str.lower()

raw_sales.loc[
    indexes[40:80],
    "Category"
] = raw_sales.loc[
    indexes[40:80],
    "Category"
].str.upper()

raw_sales.loc[
    indexes[80:],
    "Sales_Channel"
] = raw_sales.loc[
    indexes[80:],
    "Sales_Channel"
].str.lower()

# 4. Extra whitespace
indexes = np.random.choice(
    raw_sales.index,
    100,
    replace=False
)

raw_sales.loc[
    indexes,
    "Product"
] = raw_sales.loc[
    indexes,
    "Product"
].astype(str).apply(
    lambda x: "  " + x + " "
)

# 5. Invalid dates
bad_date_indexes = np.random.choice(
    raw_sales.index,
    12,
    replace=False
)

raw_sales.loc[
    bad_date_indexes[:6],
    "Order_Date"
] = "2026-13-40"

raw_sales.loc[
    bad_date_indexes[6:],
    "Order_Date"
] = "not_a_date"

# 6. Invalid numerical values
anomaly_indexes = np.random.choice(
    raw_sales.index,
    10,
    replace=False
)

raw_sales.loc[
    anomaly_indexes[:5],
    "Quantity"
] = 0

raw_sales.loc[
    anomaly_indexes[5:],
    "Unit_Price"
] = -50

# Shuffle the dataset
raw_sales = raw_sales.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

# ============================================
# DISPLAY RESULTS
# ============================================

print("RAW DATASET CREATED")
print("=" * 50)

print(f"Rows: {raw_sales.shape[0]:,}")
print(f"Columns: {raw_sales.shape[1]}")

print("\nFirst 5 records:")
display(raw_sales.head())

print("\nColumn names:")
print(raw_sales.columns.tolist())
```

## 2. Data Cleaning

After profiling the raw dataset, the next step is to clean the data before performing any business analysis.

The goal is not simply to remove rows. Each cleaning decision should have a clear reason and should preserve valid business information wherever possible.

### Cleaning Strategy

The following rules will be applied:

1. Remove duplicate transactions using `Order_ID`.
2. Convert the order date to a proper datetime format.
3. Remove transactions with invalid dates because they cannot be reliably analyzed over time.
4. Remove records with zero or negative quantities.
5. Remove records with zero or negative unit prices.
6. Standardize text fields by removing unnecessary whitespace and correcting inconsistent capitalization.
7. Replace missing customer and categorical information with `"Unknown"` where appropriate.
8. Treat a missing discount as no discount (`0%`) because the transaction does not indicate that a discount was applied.
9. Recalculate sales and profit from the cleaned transaction values to ensure financial calculations are internally consistent.
10. Create additional analytical fields that will be useful for later analysis.

Importantly, the original `raw_sales` dataset will not be modified. A separate `clean_sales` dataset will be created so that the cleaning process remains reproducible and auditable.


```python
# ============================================
# STEP 4: DATA CLEANING
# ============================================

# Keep the original raw dataset unchanged
clean_sales = raw_sales.copy()

print("Starting data cleaning...")
print(f"Initial rows: {len(clean_sales):,}")

# --------------------------------------------
# 1. Remove duplicate transactions
# --------------------------------------------

before_duplicates = len(clean_sales)

clean_sales = clean_sales.drop_duplicates(
    subset="Order_ID",
    keep="first"
)

duplicates_removed = before_duplicates - len(clean_sales)

print(f"Duplicate transactions removed: {duplicates_removed:,}")


# --------------------------------------------
# 2. Standardize text fields
# --------------------------------------------

text_columns = [
    "Customer_Name",
    "Province",
    "City",
    "Category",
    "Payment_Method",
    "Sales_Channel"
]

for column in text_columns:
    clean_sales[column] = (
        clean_sales[column]
        .astype("string")
        .str.strip()
        .str.title()
    )

# Product names only need whitespace cleanup
clean_sales["Product"] = (
    clean_sales["Product"]
    .astype("string")
    .str.strip()
)


# --------------------------------------------
# 3. Convert dates
# --------------------------------------------

clean_sales["Order_Date"] = pd.to_datetime(
    clean_sales["Order_Date"],
    errors="coerce"
)

invalid_dates = clean_sales["Order_Date"].isna().sum()

print(f"Invalid dates identified: {invalid_dates:,}")


# --------------------------------------------
# 4. Convert numerical columns
# --------------------------------------------

numeric_columns = [
    "Quantity",
    "Unit_Price",
    "Discount",
    "Sales",
    "Cost",
    "Profit"
]

for column in numeric_columns:
    clean_sales[column] = pd.to_numeric(
        clean_sales[column],
        errors="coerce"
    )


# --------------------------------------------
# 5. Remove invalid dates
# --------------------------------------------

before_dates = len(clean_sales)

clean_sales = clean_sales[
    clean_sales["Order_Date"].notna()
]

print(
    f"Rows removed because of invalid dates: "
    f"{before_dates - len(clean_sales):,}"
)


# --------------------------------------------
# 6. Remove invalid quantities
# --------------------------------------------

before_quantity = len(clean_sales)

clean_sales = clean_sales[
    clean_sales["Quantity"] > 0
]

print(
    f"Rows removed because of invalid quantity: "
    f"{before_quantity - len(clean_sales):,}"
)


# --------------------------------------------
# 7. Remove invalid unit prices
# --------------------------------------------

before_price = len(clean_sales)

clean_sales = clean_sales[
    clean_sales["Unit_Price"] > 0
]

print(
    f"Rows removed because of invalid unit price: "
    f"{before_price - len(clean_sales):,}"
)


# --------------------------------------------
# 8. Handle missing values
# --------------------------------------------

clean_sales["Customer_Name"] = (
    clean_sales["Customer_Name"]
    .fillna("Unknown")
)

clean_sales["Province"] = (
    clean_sales["Province"]
    .fillna("Unknown")
)

clean_sales["City"] = (
    clean_sales["City"]
    .fillna("Unknown")
)

clean_sales["Payment_Method"] = (
    clean_sales["Payment_Method"]
    .fillna("Unknown")
)

clean_sales["Discount"] = (
    clean_sales["Discount"]
    .fillna(0)
)


# --------------------------------------------
# 9. Validate discount values
# --------------------------------------------

clean_sales = clean_sales[
    (clean_sales["Discount"] >= 0) &
    (clean_sales["Discount"] <= 1)
]


# --------------------------------------------
# 10. Recalculate financial metrics
# --------------------------------------------

clean_sales["Sales"] = (
    clean_sales["Quantity"]
    * clean_sales["Unit_Price"]
    * (1 - clean_sales["Discount"])
)

clean_sales["Sales"] = clean_sales["Sales"].round(2)

clean_sales["Profit"] = (
    clean_sales["Sales"]
    - clean_sales["Cost"]
)

clean_sales["Profit"] = clean_sales["Profit"].round(2)


# --------------------------------------------
# 11. Create analytical fields
# --------------------------------------------

clean_sales["Profit_Margin"] = np.where(
    clean_sales["Sales"] != 0,
    clean_sales["Profit"] / clean_sales["Sales"],
    0
)

clean_sales["Profit_Margin"] = (
    clean_sales["Profit_Margin"].round(4)
)

clean_sales["Year"] = (
    clean_sales["Order_Date"].dt.year
)

clean_sales["Month"] = (
    clean_sales["Order_Date"].dt.to_period("M")
    .astype(str)
)

clean_sales["Month_Name"] = (
    clean_sales["Order_Date"]
    .dt.strftime("%B")
)


# --------------------------------------------
# 12. Sort the cleaned dataset
# --------------------------------------------

clean_sales = clean_sales.sort_values(
    "Order_Date"
).reset_index(drop=True)


# --------------------------------------------
# FINAL RESULT
# --------------------------------------------

print("\n" + "=" * 50)
print("DATA CLEANING COMPLETE")
print("=" * 50)

print(f"Final rows: {len(clean_sales):,}")
print(f"Final columns: {len(clean_sales.columns)}")

print("\nCleaned dataset preview:")
display(clean_sales.head())

print("\nRemaining missing values:")
display(
    clean_sales.isna().sum()
    .sort_values(ascending=False)
    .head(10)
)
```

## 3. Data Validation

After cleaning the dataset, I will validate the results before proceeding with exploratory analysis.

Data validation is an important part of the analytics workflow because a dataset can appear clean while still containing logical or business-rule errors.

The validation checks will confirm that:

* Order IDs are unique.
* Required fields contain no missing values.
* Dates fall within the expected period.
* Quantities are positive.
* Unit prices are positive.
* Discounts fall between 0% and 100%.
* Sales calculations are consistent with quantity, price and discount.
* Profit calculations are consistent with sales and cost.
* Profit margins are within a reasonable range.

These checks provide confidence that the cleaned dataset is suitable for further analysis and dashboard development.


```python
# ============================================
# STEP 5: DATA VALIDATION
# ============================================

print("DATA VALIDATION")
print("=" * 60)


# --------------------------------------------
# 1. Check duplicate Order IDs
# --------------------------------------------

duplicate_orders = clean_sales["Order_ID"].duplicated().sum()

print("\n1. Duplicate Order IDs")
print(f"Duplicate Order IDs found: {duplicate_orders}")


# --------------------------------------------
# 2. Check missing values
# --------------------------------------------

total_missing = clean_sales.isna().sum().sum()

print("\n2. Missing Values")
print(f"Total missing values: {total_missing}")


# --------------------------------------------
# 3. Check date range
# --------------------------------------------

min_date = clean_sales["Order_Date"].min()
max_date = clean_sales["Order_Date"].max()

print("\n3. Date Range")
print(f"Minimum date: {min_date.date()}")
print(f"Maximum date: {max_date.date()}")


# --------------------------------------------
# 4. Check quantities
# --------------------------------------------

invalid_quantity = (
    clean_sales["Quantity"] <= 0
).sum()

print("\n4. Quantity Validation")
print(f"Non-positive quantities: {invalid_quantity}")


# --------------------------------------------
# 5. Check unit prices
# --------------------------------------------

invalid_price = (
    clean_sales["Unit_Price"] <= 0
).sum()

print("\n5. Unit Price Validation")
print(f"Non-positive unit prices: {invalid_price}")


# --------------------------------------------
# 6. Check discounts
# --------------------------------------------

invalid_discount = (
    (clean_sales["Discount"] < 0) |
    (clean_sales["Discount"] > 1)
).sum()

print("\n6. Discount Validation")
print(f"Invalid discounts: {invalid_discount}")


# --------------------------------------------
# 7. Validate Sales calculation
# --------------------------------------------

expected_sales = (
    clean_sales["Quantity"]
    * clean_sales["Unit_Price"]
    * (1 - clean_sales["Discount"])
)

sales_difference = (
    clean_sales["Sales"] - expected_sales
).abs()

sales_errors = (
    sales_difference > 0.01
).sum()

print("\n7. Sales Calculation Validation")
print(f"Sales calculation errors: {sales_errors}")


# --------------------------------------------
# 8. Validate Profit calculation
# --------------------------------------------

expected_profit = (
    clean_sales["Sales"]
    - clean_sales["Cost"]
)

profit_difference = (
    clean_sales["Profit"] - expected_profit
).abs()

profit_errors = (
    profit_difference > 0.01
).sum()

print("\n8. Profit Calculation Validation")
print(f"Profit calculation errors: {profit_errors}")


# --------------------------------------------
# 9. Validate Profit Margin
# --------------------------------------------

invalid_margin = (
    (clean_sales["Profit_Margin"] < -1) |
    (clean_sales["Profit_Margin"] > 1)
).sum()

print("\n9. Profit Margin Validation")
print(f"Invalid profit margins: {invalid_margin}")


# --------------------------------------------
# 10. Create validation summary
# --------------------------------------------

validation_results = pd.DataFrame({
    "Validation_Check": [
        "Duplicate Order IDs",
        "Missing Values",
        "Non-positive Quantities",
        "Non-positive Unit Prices",
        "Invalid Discounts",
        "Sales Calculation Errors",
        "Profit Calculation Errors",
        "Invalid Profit Margins"
    ],

    "Issues_Found": [
        duplicate_orders,
        total_missing,
        invalid_quantity,
        invalid_price,
        invalid_discount,
        sales_errors,
        profit_errors,
        invalid_margin
    ]
})

validation_results["Status"] = np.where(
    validation_results["Issues_Found"] == 0,
    "PASS",
    "REVIEW"
)

print("\n" + "=" * 60)
print("VALIDATION SUMMARY")
print("=" * 60)

display(validation_results)
```

## Data Validation Summary

The cleaned dataset was subjected to a series of quality and consistency checks.

The validation process confirms whether the dataset satisfies the defined business rules before analysis begins.

A validation status of **PASS** indicates that no issues were identified for that particular check.

This step is important because reliable analysis depends on reliable data. By validating the cleaned dataset before creating visualizations and KPIs, I can reduce the risk of presenting incorrect business conclusions.

## 5. Interactive Business Explorer

To make the analysis more useful for decision-making, I will create an interactive business explorer.

Users will be able to filter the dataset by:

* Province
* Product Category
* Sales Channel

The filtered results will update key performance indicators and visualizations.

This approach demonstrates how Python can be used not only for analysis, but also to create interactive analytical tools that allow users to explore business data without manually modifying the underlying dataset.


```python
# ============================================
# STEP 15: INTERACTIVE BUSINESS EXPLORER
# COLAB VERSION
# ============================================

import ipywidgets as widgets
from IPython.display import display, clear_output


# --------------------------------------------
# FILTER OPTIONS
# --------------------------------------------

province_options = ["All"] + sorted(
    clean_sales["Province"].dropna().unique().tolist()
)

category_options = ["All"] + sorted(
    clean_sales["Category"].dropna().unique().tolist()
)

channel_options = ["All"] + sorted(
    clean_sales["Sales_Channel"].dropna().unique().tolist()
)


# --------------------------------------------
# CREATE DROPDOWNS
# --------------------------------------------

province_dropdown = widgets.Dropdown(
    options=province_options,
    value="All",
    description="Province:"
)

category_dropdown = widgets.Dropdown(
    options=category_options,
    value="All",
    description="Category:"
)

channel_dropdown = widgets.Dropdown(
    options=channel_options,
    value="All",
    description="Channel:"
)


# --------------------------------------------
# OUTPUT AREA
# --------------------------------------------

output = widgets.Output()


# --------------------------------------------
# DASHBOARD FUNCTION
# --------------------------------------------

def update_dashboard(change=None):

    # Get current filter selections
    province = province_dropdown.value
    category = category_dropdown.value
    channel = channel_dropdown.value

    filtered = clean_sales.copy()

    # Apply Province filter
    if province != "All":
        filtered = filtered[
            filtered["Province"] == province
        ]

    # Apply Category filter
    if category != "All":
        filtered = filtered[
            filtered["Category"] == category
        ]

    # Apply Channel filter
    if channel != "All":
        filtered = filtered[
            filtered["Sales_Channel"] == channel
        ]

    # ----------------------------------------
    # CLEAR PREVIOUS OUTPUT
    # ----------------------------------------

    with output:

        clear_output(wait=True)

        # ------------------------------------
        # Check whether data exists
        # ------------------------------------

        if filtered.empty:

            print("No transactions match the selected filters.")

            return

        # ------------------------------------
        # KPI CALCULATIONS
        # ------------------------------------

        total_sales = filtered["Sales"].sum()

        total_profit = filtered["Profit"].sum()

        total_orders = filtered["Order_ID"].nunique()

        total_units = filtered["Quantity"].sum()

        average_order_value = (
            total_sales / total_orders
        )

        profit_margin = (
            total_profit / total_sales
        )

        # ------------------------------------
        # DISPLAY FILTER SELECTION
        # ------------------------------------

        print("=" * 65)
        print("INTERACTIVE BUSINESS EXPLORER")
        print("=" * 65)

        print(
            f"\nFilters:"
            f"\nProvince: {province}"
            f"\nCategory: {category}"
            f"\nChannel: {channel}"
        )

        print("\n" + "-" * 65)

        # ------------------------------------
        # DISPLAY KPIs
        # ------------------------------------

        print(f"Total Sales:         ${total_sales:,.2f}")
        print(f"Total Profit:        ${total_profit:,.2f}")
        print(f"Total Orders:        {total_orders:,}")
        print(f"Units Sold:          {total_units:,}")
        print(f"Average Order Value: ${average_order_value:,.2f}")
        print(f"Profit Margin:       {profit_margin:.2%}")

        print("\n" + "-" * 65)

        # ------------------------------------
        # PRODUCT PERFORMANCE
        # ------------------------------------

        product_summary = (
            filtered
            .groupby("Product", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
            .sort_values(
                "Sales",
                ascending=False
            )
            .head(10)
        )

        fig1 = px.bar(
            product_summary.sort_values("Sales"),
            x="Sales",
            y="Product",
            orientation="h",
            title="Top Products by Sales",
            labels={
                "Sales": "Sales ($)",
                "Product": "Product"
            }
        )

        fig1.show()

        # ------------------------------------
        # MONTHLY PERFORMANCE
        # ------------------------------------

        monthly = (
            filtered
            .groupby("Month", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
        )

        fig2 = px.line(
            monthly,
            x="Month",
            y=["Sales", "Profit"],
            markers=True,
            title="Monthly Sales and Profit",
            labels={
                "value": "Amount ($)",
                "Month": "Month",
                "variable": "Metric"
            }
        )

        fig2.update_layout(
            hovermode="x unified"
        )

        fig2.show()

        # ------------------------------------
        # SHOW FILTERED DATA SAMPLE
        # ------------------------------------

        print("\nFiltered Transaction Sample:")

        display(
            filtered[
                [
                    "Order_ID",
                    "Order_Date",
                    "Province",
                    "City",
                    "Product",
                    "Category",
                    "Sales",
                    "Profit",
                    "Sales_Channel"
                ]
            ].head(10)
        )


# --------------------------------------------
# CONNECT DROPDOWNS TO FUNCTION
# --------------------------------------------

province_dropdown.observe(
    update_dashboard,
    names="value"
)

category_dropdown.observe(
    update_dashboard,
    names="value"
)

channel_dropdown.observe(
    update_dashboard,
    names="value"
)


# --------------------------------------------
# DISPLAY FILTERS
# --------------------------------------------

display(
    widgets.VBox([
        province_dropdown,
        category_dropdown,
        channel_dropdown
    ])
)

display(output)


# --------------------------------------------
# RUN INITIAL DASHBOARD
# --------------------------------------------

update_dashboard()
```

```python
# ============================================
# STEP 18: EXECUTIVE KPI DASHBOARD
# ============================================

# Calculate executive KPIs

kpi_sales = clean_sales["Sales"].sum()
kpi_profit = clean_sales["Profit"].sum()
kpi_orders = clean_sales["Order_ID"].nunique()
kpi_aov = kpi_sales / kpi_orders
kpi_margin = kpi_profit / kpi_sales

kpi_customers = clean_sales["Customer_ID"].nunique()

kpi_repeat_customers = (
    clean_sales
    .groupby("Customer_ID")["Order_ID"]
    .nunique()
    .gt(1)
    .sum()
)

kpi_repeat_rate = kpi_repeat_customers / kpi_customers


# Display KPIs
print("=" * 75)
print("EXECUTIVE KPI DASHBOARD")
print("=" * 75)

print(f"\nTotal Sales:           ${kpi_sales:,.2f}")
print(f"Total Profit:          ${kpi_profit:,.2f}")
print(f"Total Orders:          {kpi_orders:,}")
print(f"Average Order Value:   ${kpi_aov:,.2f}")
print(f"Profit Margin:         {kpi_margin:.2%}")
print(f"Repeat Customer Rate:  {kpi_repeat_rate:.2%}")

print("\n" + "=" * 75)
```

This dashboard brings the analysis together into an interactive management view.

```python
# ============================================
# STEP 19: INTERACTIVE MANAGEMENT DASHBOARD
# ============================================

import plotly.express as px
import ipywidgets as widgets
from IPython.display import display, clear_output


# --------------------------------------------
# FILTER OPTIONS
# --------------------------------------------

province_options = ["All"] + sorted(
    clean_sales["Province"].unique().tolist()
)

category_options = ["All"] + sorted(
    clean_sales["Category"].unique().tolist()
)

channel_options = ["All"] + sorted(
    clean_sales["Sales_Channel"].unique().tolist()
)


# --------------------------------------------
# CREATE FILTERS
# --------------------------------------------

province_filter = widgets.Dropdown(
    options=province_options,
    value="All",
    description="Province:"
)

category_filter = widgets.Dropdown(
    options=category_options,
    value="All",
    description="Category:"
)

channel_filter = widgets.Dropdown(
    options=channel_options,
    value="All",
    description="Channel:"
)

dashboard_output = widgets.Output()


# --------------------------------------------
# DASHBOARD FUNCTION
# --------------------------------------------

def update_management_dashboard(change=None):

    province = province_filter.value
    category = category_filter.value
    channel = channel_filter.value

    filtered = clean_sales.copy()

    # Apply filters
    if province != "All":
        filtered = filtered[
            filtered["Province"] == province
        ]

    if category != "All":
        filtered = filtered[
            filtered["Category"] == category
        ]

    if channel != "All":
        filtered = filtered[
            filtered["Sales_Channel"] == channel
        ]

    with dashboard_output:

        clear_output(wait=True)

        if filtered.empty:
            print("No data matches the selected filters.")
            return

        # ------------------------------------
        # KPI CALCULATIONS
        # ------------------------------------

        sales = filtered["Sales"].sum()
        profit = filtered["Profit"].sum()
        orders = filtered["Order_ID"].nunique()
        units = filtered["Quantity"].sum()

        aov = sales / orders
        margin = profit / sales

        # ------------------------------------
        # KPI DISPLAY
        # ------------------------------------

        print("=" * 80)
        print("INTERACTIVE MANAGEMENT DASHBOARD")
        print("=" * 80)

        print(
            f"\nFilters:"
            f"\nProvince: {province}"
            f"\nCategory: {category}"
            f"\nChannel: {channel}"
        )

        print("\n" + "-" * 80)

        print(f"Total Sales:          ${sales:,.2f}")
        print(f"Total Profit:         ${profit:,.2f}")
        print(f"Total Orders:         {orders:,}")
        print(f"Units Sold:           {units:,}")
        print(f"Average Order Value:  ${aov:,.2f}")
        print(f"Profit Margin:        {margin:.2%}")

        print("\n" + "-" * 80)


        # ------------------------------------
        # MONTHLY PERFORMANCE
        # ------------------------------------

        monthly = (
            filtered
            .groupby("Month", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
        )

        fig_monthly = px.line(
            monthly,
            x="Month",
            y=["Sales", "Profit"],
            markers=True,
            title="Monthly Sales and Profit",
            labels={
                "value": "Amount ($)",
                "Month": "Month",
                "variable": "Metric"
            }
        )

        fig_monthly.update_layout(
            hovermode="x unified"
        )

        fig_monthly.show()


        # ------------------------------------
        # CATEGORY PERFORMANCE
        # ------------------------------------

        category = (
            filtered
            .groupby("Category", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
            .sort_values("Sales", ascending=True)
        )

        fig_category = px.bar(
            category,
            x="Sales",
            y="Category",
            orientation="h",
            title="Sales by Category",
            labels={
                "Sales": "Sales ($)",
                "Category": "Category"
            }
        )

        fig_category.show()


        # ------------------------------------
        # PROVINCE PERFORMANCE
        # ------------------------------------

        province = (
            filtered
            .groupby("Province", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
            .sort_values("Sales", ascending=True)
        )

        fig_province = px.bar(
            province,
            x="Sales",
            y="Province",
            orientation="h",
            title="Sales by Province",
            labels={
                "Sales": "Sales ($)",
                "Province": "Province"
            }
        )

        fig_province.show()


        # ------------------------------------
        # SALES CHANNEL
        # ------------------------------------

        channel = (
            filtered
            .groupby("Sales_Channel", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
            .sort_values("Sales", ascending=True)
        )

        fig_channel = px.bar(
            channel,
            x="Sales",
            y="Sales_Channel",
            orientation="h",
            title="Sales by Channel",
            labels={
                "Sales": "Sales ($)",
                "Sales_Channel": "Sales Channel"
            }
        )

        fig_channel.show()


        # ------------------------------------
        # TOP 10 PRODUCTS
        # ------------------------------------

        products = (
            filtered
            .groupby("Product", as_index=False)
            .agg(
                Sales=("Sales", "sum"),
                Profit=("Profit", "sum")
            )
            .sort_values("Sales", ascending=False)
            .head(10)
            .sort_values("Sales", ascending=True)
        )

        fig_products = px.bar(
            products,
            x="Sales",
            y="Product",
            orientation="h",
            title="Top 10 Products by Sales",
            labels={
                "Sales": "Sales ($)",
                "Product": "Product"
            },
            hover_data=["Profit"]
        )

        fig_products.show()


# --------------------------------------------
# CONNECT FILTERS TO DASHBOARD
# --------------------------------------------

province_filter.observe(
    update_management_dashboard,
    names="value"
)

category_filter.observe(
    update_management_dashboard,
    names="value"
)

channel_filter.observe(
    update_management_dashboard,
    names="value"
)


# --------------------------------------------
# DISPLAY DASHBOARD
# --------------------------------------------

display(
    widgets.HBox([
        province_filter,
        category_filter,
        channel_filter
    ])
)

display(dashboard_output)

update_management_dashboard()
```
